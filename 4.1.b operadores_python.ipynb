# 🐍 Práctica: Operadores en Python

## 1️⃣ Operadores aritméticos

Se utilizan para realizar operaciones matemáticas básicas con números.

```python
# ✅ Soluciones
# 1
x = 8
y = 5

print("Suma:", x + y)        
print("Resta:", x - y)       
print("Multiplicación:", x * y)
print("División:", x / y)
# 2
base = 12
altura = 8

print("el area es: ", (base * altura)/2)
# 3
print("resto: ", 45 % 7)
# 4
print(2 ** 10)
# 5
a = 7 + 9 + 10
print (a / 3)
```

## 2️⃣ Operadores de asignación

Permiten asignar valores a variables o modificar su contenido.

```python
# ✅ Soluciones
# 1
n = 50
n += 25
print(n)
# 2
n -= 10
print(n)
# 3
n = n * 3
print(n)
# 4
n = n / 5
print(n)
# 5
precio = 120 
precio *= 0.8
print (precio)
```

## 3️⃣ Operadores de comparación

Sirven para comparar valores. Devuelven `True` (verdadero) o `False` (falso).

```python
# ✅ Soluciones
# 1
print(8 == 8)
# 2
print(12 < 7)
# 3
print(9 >=9)
# 4
print("hola" == "adios")
# 5
print(100 != 101)
```

## 4️⃣ Operadores lógicos

Se utilizan para combinar comparaciones o expresiones booleanas.

- `and`: True si **ambas** condiciones son verdaderas.
- `or`: True si **una o ambas** condiciones son verdaderas.
- `not`: invierte el valor lógico.

```python
# ✅ Soluciones
# 1
a = 7
b = 5
c = 10
print((a > b) and (a < c)) 
# 2
print((a < b) or (b == 5))
# 3
print(not (a == 10))  
# 4
print(3 > 2)
print(10 < 5)
# 5
resultado = (1 and not 4) or 10
```

## 5️⃣ Operadores de pertenencia (`in`, `not in`)

Comprueban si un elemento está (o no está) dentro de una lista, cadena o colección.

```python
# ✅ Soluciones
# 1
colors = ['red', 'green', 'blue']
print("blue" in colors)
# 2
print('yellow' not in colors)
# 3
cadena = 'python'
print('p' in cadena) 
# 4
print('z' not in cadena) 
# 5
colors.append('purple')
print('purple' in colors)  
```

## 6️⃣ Operadores de identidad (`is`, `is not`)

Comprueban si dos variables **apuntan al mismo objeto** en memoria.

```python
# ✅ Soluciones
# 1
promedio = (4 + 6 + 8) / 3  
resultado = promedio * 2     
print(resultado)            
# 2
cadena = 'python'
print('o' in cadena)  
# 3
x = 5
x += 5  
x += 5  
x += 5 
print(x)  
# 4
numero = 15

# Opción 1: Usando el operador 'and'
print(numero > 10 and numero < 20)  
# 5
lista_a = [1, 2, 3]
lista_b = [1, 2, 3]  

# Comprobaciones
print(lista_a == lista_b)      
print(lista_a is not lista_b)  
```
